## Exercise on Single Variable linear regression
## Problem : predict the per capita of canada in year 2020

```python
# neccesary modules
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
# mount the drive for dataset
from google.colab import drive
drive.mount("/content/drive")
```

```python
# load the file
df = pd.read_csv("/content/drive/MyDrive/Datasets/canada_per_capita_income.csv")
df.head(5)
```

```python
# plotting the scatter chart
%matplotlib inline
plt.xlabel("year")
plt.ylabel("per capita income(USD)")
plt.scatter(df["year"],df["per capita income (US$)"],color="red",marker="+")
```

```python
# implement linear regression
reg = linear_model.LinearRegression()
reg.fit(df[["year"]],df["per capita income (US$)"])
```

```python
# predict the income
reg.predict([[2020]])
```

```python
# plot the predicted graph
%matplotlib inline
plt.xlabel("year")
plt.ylabel("per capita income (USD$)")
plt.scatter(df["year"],df["per capita income (US$)"],color="red",marker="+")
plt.plot(df[["year"]],reg.predict(df[["year"]]),color="blue")
```
