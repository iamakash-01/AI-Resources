## Single Independent variable linear regression formula

Screenshot from 2025-01-26 18-34-25.png

```python
# plotting a scatter plot
%matplotlib inline
plt.xlabel("sq(in feet)")
plt.ylabel("price")
plt.scatter(df["area"], df["price"], color = "red", marker = "+")
```

```python
# implementing linear regression
reg = linear_model.LinearRegression()
reg.fit(df[["area"]],df["price"])
```

```python
# ready for prediction
# predict requires 2d array
reg.predict([[3300]])
```

```python
# to check the values of cofficient and intercept
reg.coef_
```

```python
reg.intercept_
```

```python
# for 5000 sq ft
reg.predict([[5000]])
```

```python
# plot the chart with the predictions
%matplotlib inline
plt.xlabel("area(in sq ft)")
plt.ylabel("price(in USD)")
plt.scatter(df["area"], df["price"], color = "red", marker = "+")
plt.plot(df["area"],reg.predict(df[["area"]]), color = "blue")
```
