## Problem : predict the sales of TV according to the TV price for 300.6 US$

```python
# neccesary modules
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
# load the file
df = pd.read_csv("/content/drive/MyDrive/Datasets/tvmarketing.csv")
df.head(5)
```

```python
# plot the chart
%matplotlib inline
plt.xlabel("TV Price")
plt.ylabel("Sales")
plt.scatter(df["TV"],df["Sales"],color="red",marker="+")
```

```python
# implement Linear Regression
reg = linear_model.LinearRegression()
reg.fit(df[["TV"]],df["Sales"])
```

```python
# predict
reg.predict([[300.6]])
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
# verify
m = reg.coef_
b = reg.intercept_
x = 300.6
y = m*x + b
y
```

```python
# plot the linear chart
%matplotlib inline
plt.xlabel("TV Price")
plt.ylabel("Sales")
plt.scatter(df["TV"],df["Sales"],color="red",marker="+")
plt.plot(df[["TV"]],reg.predict(df[["TV"]]),color = "blue")
```
