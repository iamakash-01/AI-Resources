# Email Spam Classifier 

## 1. Using Naive Bayes MultinomialNB classifier

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report

# load the file
df = pd.read_csv("/kaggle/input/spam-dataset/spam.csv")
# df.head

# create new column for numerics
df["spam"] = df["Category"].apply(lambda x : 1 if x == "spam" else 0)
# df["spam"]

# split the dataset
X_train,X_test,y_train,y_test = train_test_split(df["Message"],df["spam"],test_size = 0.2)

# use pipeline
classifier = Pipeline([
    ("vectorizer", CountVectorizer()),
    ("nb", MultinomialNB())
])

# fit the dataset
classifier.fit(X_train,y_train)

# predict the data
predict = classification_report(y_test,classifier.predict(X_test))
print(predict)
```

```python
# checking with a sample email
# Here 0 means Ham and 1 means spam
emails = [
    "Hey Mohan, can we together for a football match tonight?",
    "Upto 20% discount on parking, exclusive offer just for you. Don't miss this reward!"
]
classifier.fit(X_train,y_train)
sample_emails = classifier.predict(emails)
print(sample_emails)
```

## 2. Using RandomForest classifier

```python
# classifier
from sklearn.ensemble import RandomForestClassifier
classifier = Pipeline([
    ("vectorizer", CountVectorizer()),
    ("rf", RandomForestClassifier(n_estimators = 50, criterion = "entropy"))
])

# fit the dataset
classifier.fit(X_train,y_train)

# prediction
print(classification_report(y_test,classifier.predict(X_test)))
```
