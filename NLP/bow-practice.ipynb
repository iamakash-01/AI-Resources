# Movie Sentiment classification using Bag-Of-Word

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/kaggle/input/movie-dataset/movies_sentiment_data.csv")
df.head
```

```python
df.shape
```

```python
# checks the count
df["sentiment"].value_counts()
```

```python
df["Category"] = df["sentiment"].apply(lambda x : 1 if x == "positive" else 0)
```

```python
df["Category"]
```

```python
# split the dataset
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df["review"],df["Category"],test_size = 0.2)
```

```python
X_train.shape
```

```python
X_test.shape
```

# Exercise 1 : fetch the classification report

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report
from sklearn.pipeline import Pipeline

clf = Pipeline([
    ("vectorizer", CountVectorizer()),
    ("nb", MultinomialNB())
])

clf.fit(X_train,y_train)
```

```python
predict = classification_report(y_test, clf.predict(X_test))
print(predict)
```
