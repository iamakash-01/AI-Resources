## Problem : predict the salary for 20 year of experience

```python
# load the drive
from google.colab import drive
drive.mount("/content/drive")
```

```python
# neccesary modules
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
# load the file
df = pd.read_csv("/content/drive/MyDrive/Datasets/Salary_dataset.csv")
df.head(5)
```

```python
# plot the unpredicted chart
%matplotlib inline
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.scatter(df["YearsExperience"],df["Salary"],color="red",marker="+")
```

```python
# simple linear regression implementaion
reg = linear_model.LinearRegression()
reg.fit(df[["YearsExperience"]],df["Salary"])
```

```python
# prediction
reg.predict([[20]])
```

```python
# verify the value of cofficient and intercept
print(reg.coef_)
print(reg.intercept_)
```

```python
# verify
m = reg.coef_
b = reg.intercept_
x = 20
y = (m*x) + b
y
```

```python
# plot the predicted chart
%matplotlib inline
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.scatter(df["YearsExperience"],df["Salary"],color="red",marker="+")
plt.plot(df[["YearsExperience"]],reg.predict(df[["YearsExperience"]]),color="blue")
```

## Solution : The salary for 20 years of experience person is 213847.45039562 based on the dataset given
